```python
x = [1, 2, 3, 4]
f = [2, 3, 4, 5]
fx = 0
for i in range(len(x)):
  fx = fx+x[i]*f[i]

mean = fx/sum(f)
print(mean)
```

```python
x = [1, 2, 3, 4, 5]
sm = 0
mean = sum(x)/len(x)
for i in range(len(x)):
  sm = sm + (x[i]-mean)**2
rmse = (sm/len(x))**0.5
print(rmse)
```

```python
data = [5, 8, None, 7, 10]
s = 0
count = 0
for x in data:
  if x is not None:
    s+=x
    count+=1
mean_val = s/count
imputed_data = data[:]   #copy of original list
for i in range(len(imputed_data)):
    if imputed_data[i] is None:
        imputed_data[i] = mean_val   # only update where None

print("original data:", data)
print("imputed data:", imputed_data)

```

```python
x = [2,3,4,5,6,7]
x.sort()
n = len(x)
if n % 2 ==0:
  mid1 = n//2
  mid2 = mid1-1
  median = (x[mid1]+x[mid2])/2
else:
  mid1 = n//2
  median = x[mid1]
print(median)
```
